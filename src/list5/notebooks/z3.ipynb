```python
import numpy as np
import matplotlib.pyplot as plt
from pymoo.algorithms.moo.nsga2 import NSGA2
from pymoo.problems import get_problem
from pymoo.optimize import minimize
from pymoo.visualization.scatter import Scatter

problems = [
    ('ZDT1', get_problem("zdt1")),
    ('ZDT2', get_problem("zdt2")),
    ('ZDT3', get_problem("zdt3"))
]

algorithm = NSGA2(
    pop_size=100,
)

for name, problem in problems:
    print(f"Solving {name}...")
    res = minimize(problem, algorithm, ('n_gen', 200), seed=42,)

    plot = Scatter(title=f"Pareto Front for {name}")
    plot.add(problem.pareto_front(), plot_type="line", color="green", alpha=0.7, label="True Front")
    plot.add(res.F, color="red", label="NSGA-II Solution")
    plot.show()
    plt.show()
```
